# Objective Function

```python
def multi_objective(trial, average_vs, average_fxs, average_fys, des_speeds, durations):

    #delay = trial.suggest_float('delay', 0.045, 0.0706)
    delay = trial.suggest_float('delay', 0.06, 0.0706)
    
    K_fy = trial.suggest_float('K_fy', -0.0004, 0)
    K_fx = trial.suggest_float('K_fx', 0.0000001, 0.2)
    Kp = trial.suggest_float('Kp', 0.001, 20)
    Kd = trial.suggest_float('Kd', -0.05, 0)

    n_trials = len(average_vs)
    all_errors = []
#    try:
    for n in range(n_trials):
          
            average_v = average_vs[n]
            des_speed = des_speeds[n]
            duration = durations[n]
            delta_t = duration / len(average_v)
            delay_samples = int(delay/delta_t)
        
        
            fy = [0 if i  < -10 else i for i in average_fys[n]]
            fx = [i for i in average_fxs[n]]
    #    average_v = [des_speed if fy[i]  < 10 else average_v[i] for i in range(len(fy))]


            start_index = 90
    
            speeds = np.zeros(100)
            speed = des_speeds[n]#starting speed is the last speed before heel contact = 0
            speeds[start_index-1] = speed
            fx_all = deque(maxlen=delay_samples)
            fy_all = deque(maxlen=delay_samples)
            e_all = deque(maxlen=2) #with no delay, add to add delay
            e = 0


            e_all.append(e)

            for k in range(len(average_v)-1):    
            # total of 100 iterations
                i = (start_index + k) % len(average_v)
         
                if fy[i] < -10:
                    speed = des_speed
                else:
                    dfxdt = (fx[i] - fx[i-1])/delta_t

                    fx_all.append(dfxdt)

                    dfydt = (fy[i] - fy[i-1])/delta_t
                    fy_all.append(dfydt)

        #PD PART
                    e = des_speed - speed #error at node 2 is 1.2 - error at node 1
                    e_all.append(e)
                    dedt = (e_all[1] - e_all[0]) / delta_t

                    grf_part = K_fx *fx_all[0]  + K_fy * fy_all[0]



                    pd_part  = Kp*e + Kd*dedt #- Ki*edt
                    speed = speed + grf_part +pd_part #this option always computes delta speed
                speeds[i] = speed
        


            MSE = np.square(np.subtract(average_v,speeds)).mean()
            obj = MSE#np.sqrt(MSE)
            all_errors.append(obj)
  #  except:
  #     all_errors.append(20)
 
    return np.mean(all_errors)
```

```python
def final_multi(average_vs, average_fxs, average_fys, des_speeds, durations):
    delay = params['delay']
    delta_t = duration / len(average_v)
    delay_samples = int(delay/delta_t)
    K_fy = params['K_fy']
    Kp = params['Kp']
    Kd = params['Kd']
    K_fx = params['K_fx']
    n_trials = len(average_vs)
    all_speeds = []
    for n in range(n_trials):
        average_v = average_vs[n]
        des_speed = des_speeds[n]
        duration = durations[n]
        delta_t = duration / len(average_v)
        delay_samples = int(delay/delta_t)
        
        
        fy = [0 if i  < -10 else i for i in average_fys[n]]
        fx = [i for i in average_fxs[n]]
       # average_v = [des_speed if fy[i]  < 10 else average_v[i] for i in range(len(fy))]


        start_index = 90
    
        speeds = np.zeros(100)
        speed = des_speeds[n]#starting speed is the last speed before heel contact = 0
        speeds[start_index-1] = speed
        fx_all = deque(maxlen=delay_samples)
        fy_all = deque(maxlen=delay_samples)
        e_all = deque(maxlen=2) #with no delay, add to add delay
        e = 0


        e_all.append(e)

        for k in range(len(average_v)-1):    
            # total of 100 iterations
            i = (start_index + k) % len(average_v)
         
            if fy[i] < -10:
                speed = des_speed
            else:
                dfxdt = (fx[i] - fx[i-1])/delta_t

                fx_all.append(dfxdt)

                dfydt = (fy[i] - fy[i-1])/delta_t
                fy_all.append(dfydt)

        #PD PART
                e = des_speed - speed #error at node 2 is 1.2 - error at node 1
                e_all.append(e)
                dedt = (e_all[1] - e_all[0]) / delta_t

                grf_part = K_fx *fx_all[0]  + K_fy * fy_all[0]



                pd_part  = Kp*e + Kd*dedt #- Ki*edt
                speed = speed + grf_part +pd_part #this option always computes delta speed
            speeds[i] = speed


        all_speeds.append(speeds)
 
    return all_speeds
```

```python
def final(average_v, average_fx, average_fy, params, des_speed, duration):
    delay = params['delay']
    delta_t = duration / len(average_v)
    delay_samples = int(delay/delta_t)
    K_fy = params['K_fy']
    Kp = params['Kp']
    Kd = params['Kd']
    K_fx = params['K_fx']
    start_index = int(90 * len(average_v)/100)

    average_fy = [0 if i  < 1 else i for i in average_fy]
    average_fx = [i for i in average_fx]
   # average_v = [des_speed if average_fy[i]  < 10 else average_v[i] for i in range(len(average_fy))]

    fx = average_fx
    fy = average_fy
    
    speed = des_speed#starting speed is the last speed before heel contact = 0
    speeds = np.zeros(len(average_v))
    speeds[start_index-1] = speed
    fx_all = deque(maxlen=delay_samples)
    fy_all = deque(maxlen=delay_samples)
    e_all = deque(maxlen=2) #with no delay, add to add delay
    e = 0


    e_all.append(e)

    for k in range(len(average_v)-1):          # total of 100 iterations
        i = (start_index + k) % len(average_v)        # wraps back to 0 after 99
    
        if fy[i] < -10:
            speed = des_speed
        else:
            dfxdt = (fx[i] - fx[i-1])/delta_t

            fx_all.append(dfxdt)

            dfydt = (fy[i] - fy[i-1])/delta_t
            fy_all.append(dfydt)

        #PD PART
            e = des_speed - speed #error at node 2 is 1.2 - error at node 1
            e_all.append(e)
            dedt = (e_all[1] - e_all[0]) / delta_t

            grf_part = K_fx *fx_all[0]  + K_fy * fy_all[0]



            pd_part  = Kp*e + Kd*dedt #- Ki*edt
            speed = speed + grf_part +pd_part #this option always computes delta speed
        speeds[i] = speed

    return speeds
```

```python
participants = np.unique(dataset.index.id)
```

```python
loo = LeaveOneOut()
loo.get_n_splits(participants)
```

# 1) fit per speed

```python
speed = 1.8
```

```python
test_speeds = []
test_errors = []
train_errors = []
all_params = []
for i, (train_index, test_index) in enumerate(loo.split(participants)):
   # if i<8:
    #    continue
    print('fold', i)
    idx = train_index
    test_idx = test_index
    trainset = dataset.get_subset(id=list(participants[idx]), speed=[speed])
    testset = dataset.get_subset(id=list(participants[test_idx]), speed=[speed])
    vs, fys, fxs, durs, des_vs = [],[],[],[], []
    #iterate over all participants in the trainset to load the data
    for p in trainset: 
        cycles, dur = get_data(p)
       # dur = np.round(dur, 3)
        vs.append(cycles.speedL)
        fys.append(cycles.FyL)
        fxs.append(cycles.FxL)
        durs.append(dur)
        des_vs.append(p.index.speed[0])

    #optimize params
    func_multi = lambda trial: multi_objective(trial, vs, fxs, fys, des_vs, durs)
    sampler = optuna.samplers.CmaEsSampler(x0=x0, seed=0)
   # sampler = optuna.samplers.TPESampler(seed=0)
    study_multi = optuna.create_study(sampler = sampler)
    study_multi.optimize(func_multi, n_trials=4000)
    train_errors.append(study_multi.best_value)
    all_params.append(study_multi.best_params)

    #test on left out participant
    if len(testset) == 1:
        p = testset[0]
        cycles, dur = get_data(p)
      #  dur = np.round(dur, 3)
        test_speed = final(cycles.speedL, cycles.FxL, cycles.FyL, study_multi.best_params, p.index.speed[0], dur)
        error = np.sqrt(np.square(np.subtract(cycles.speedL,test_speed)).mean())
    test_errors.append(error)
    test_speeds.append(test_speed)
```
